```python
import re

df_inputs["text"] = df_inputs["text"].astype(str)

def limpar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r'\s+', ' ', texto)
    texto = texto.strip()
    return texto

df_inputs["text_limpo"] = df_inputs["text"].apply(limpar_texto)

print("Preparação concluída!")
display(df_inputs[["id", "text_limpo"]].head())
```
